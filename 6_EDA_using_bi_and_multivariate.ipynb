```python
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df =sns.load_dataset('tips')
```

1:scatterplot ( numerical - numerical)

```python
sns.scatterplot(
    data = df ,
    x = 'total_bill',
    y = 'tip',
)
```

```python

sns.scatterplot(
    data = df ,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex',
    style = 'smoker',

)
```

```python
df = sns.load_dataset('titanic')
sns.barplot(
    data = df,
    x = 'pclass',
    y = 'age',
    hue = 'sex'
)
```

boxplot ( numerical - categorical)

```python
sns.boxplot(
    data = df,
    x='sex',
    y ='age',
    hue='survived'
)
```

```python
sns.kdeplot(
    data=df[df['survived']==0],
    x = 'age'
)

sns.kdeplot(
    data = df[df['survived']==1],
    x = 'age'
)
```
