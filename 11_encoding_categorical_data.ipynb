Ordinal ecoding handling ordinal data


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df  = pd.read_csv('/content/deepseek_csv_20261006_940c98.txt')
df.columns = ["age", "review", "qualification", "purchased", "gender"]
df.sample(5)
```

```python
print(df.columns)
```

```python
df_new = df.iloc[:,1:4]
```

```python
df_new.head()
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , Y_train , Y_test = train_test_split (
df_new.iloc[:,0:2],df_new.iloc[:,-1],
 test_size = 0.2,
  random_state = 42
)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
X_train.head()
```

```python
oe = OrdinalEncoder(categories =[['excellent','good','bad'],['pg','ug','hs']] )
```

```python
oe.fit(X_train)
```

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
X_train
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
le.fit(Y_train)
Y_train = le.transform(Y_train)
Y_test = le.transform(Y_test)
```
