Loading data directly in a csv file using the url and request method

```python
import requests
from io import StringIO
URL = "https://www.kaggle.com/datasets/datascikhan/e-commerce-sales-and-customer-analytics"
headers = { "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) ..."}
req  = requests.get(URL, headers=headers)
data = StringIO(req.text)
pd.read_csv(data)
```
